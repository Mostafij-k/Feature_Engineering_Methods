```python
# Chi-square test is used to check whether there is a relationship between categorical features and the target variable.
    # It measures dependency between feature and target
    # Higher Chi-square score ⇒ more important feature
    # Works best for classification problems
```

```python
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
```

```python
# Example dataset
df = pd.read_csv("titanic.csv")

X = df[['Pclass', 'Age', 'Fare']]
y = df['Survived']
```

```python
# Handle missing values
X = X.fillna(0)
```

```python
# Scale to non-negative values
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Apply SelectKBest with Chi-square
selector = SelectKBest(score_func=chi2, k=2)
X_new = selector.fit_transform(X_scaled, y)
```

```python
# Get scores
scores = selector.scores_

print("Feature Scores:", scores)
```

```python
# Selected Feature
selected_features = selector.get_support()
print(selected_features)
```

```python
# Wrapper methods are a type of feature selection methods that involve using a predictive model(ML model) to score the combination of features.

# It create a subset of input feature from orginal feature than try ML algorithm and find r2/accuracy score. Based on r2 score it select best feature.

# They are called 'Wrapper' method because they 'wrap' this type of model based evaluation around the feature selection process.

# Step of wrapper method
    # 1. Subset Generation
    # 2. Subset Evaluation
    # 3. Stoping Criterion
```

```python
# Problems with too many features include:

# Overfitting

# High computational cost

# Lower model interpretability

# Noise affecting predictions
```

```python
# Type of Wrapper feature selection
    # Exhaustive feature selection
    # Forward selection
    # Backward elimination
    # Recursive feature elimination
```

```python
# Exhaustive feature selection/Best subset selection
    # Try out all the subset combination
    # Example:
    # f1,f2,y: try: f1-y, f2-y, f1,f2-y : find score and select best subset combination.
```

```python
# Disadvantages:
    # 1. Computational Complexity: Need 2^n-1 evaluation
    # 2. Risk of overfitting
    # 3. Require good evaluation metric

# Note: It is best for low number of feature dataset.
# It is better to use filter method than use wrapper method
```

```python
# Step-by-Step Process
# F={X1​,X2​,X3​,X4​,X5​}
# Train a model using all features.
# Remove X1......=F1 score
# Remove X2.......= f1 score
#.......
```

```python
# Sequential Forward Selection(SFS):
# Sequential Forward Selection starts with no features.
# Adds one feature at a time that improves the model performance the most.
# Start → No Features → Add one by one → Best subset
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from mlxtend.feature_selection import SequentialFeatureSelector as SFS
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
# Load Dataset
Data=pd.read_csv('BostonHousing.csv')
Data.sample(3)
```

```python
# separate the target variable
X = Data.drop("medv", axis=1)
y = Data['medv']

# split the data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

print(X_train.shape)
```

```python
# Perform StandardScalar
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
# Perform Linear regression model
model = LinearRegression()

print("training",np.mean(cross_val_score(model, X_train, y_train, cv=5, scoring='r2')))
print("testing",np.mean(cross_val_score(model, X_test, y_test, cv=5, scoring='r2')))
```

```python
# Perform backward elimination

lr = LinearRegression()

# perform backward elimination
sfs = SFS(lr, k_features='best',
           forward=True, 
           floating=False, 
           scoring='r2',cv=5)

sfs.fit(X_train, y_train)
```

```python
# Show the best feature index
sfs.k_feature_idx_
```

```python
# Now perform linear regression
X_train_sel = sfs.transform(X_train)
X_test_sel = sfs.transform(X_test)

model = LinearRegression()

print("training",np.mean(cross_val_score(model, X_train_sel, y_train, cv=5, scoring='r2')))
print("testing",np.mean(cross_val_score(model, X_test_sel, y_test, cv=5, scoring='r2')))
```

```python
# Recursive Feature Elimination:
# RFE repeatedly builds a model and removes the weakest feature until only the most useful features remain.
# It works by recursively removing the least important features and building the model again until the optimal number of features remains.
```

```python
# Recursive elimination take two argument:
# 1. Algorithm
# 2. Feature
# Than base on algorithm it select best feature
# Feature selection vary algorithm to algorithm
# DT give weight, LR give coefficient and select feature
```

```python
# RFE follows a backward selection process.

# It starts with all features, then removes the least important feature step by step.

# Process
    # Train a model using all features.
    # Measure feature importance.
    # Remove the least important feature.
    # Train the model again with remaining features.
    # Repeat until desired number of features remains.
    # This is why it is called recursive (repeated process).
```

```python
# RFE helps to:

# Remove irrelevant features

# Improve model performance

# Reduce model complexity
```

```python
# If dataset has n features

# RFE performs:

# 1st iteration → n features
# 2nd iteration → n−1 features
# 3rd iteration → n−2 features
# Until k features remain.
```

```python
# Import Data set
Wine_data=pd.read_csv('wine_data.csv')
```

```python
# Set column name
Wine_data.columns = ['Class label', 'Alcohol', 'Malic acid', 'Ash',
                   'Alcalinity of ash', 'Magnesium', 'Total phenols',
                   'Flavanoids', 'Nonflavanoid phenols', 'Proanthocyanins',
                   'Color intensity', 'Hue', 'OD280/OD315 of diluted wines',
                   'Proline']
```

```python
# Show the data set
Wine_data.head(3)
print('Class labels', np.unique(Wine_data['Class label']))
```

```python
# separate the target variable
X = Wine_data.drop('Class label', axis=1)
y = Wine_data['Class label']

# split the data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2, 
    random_state=1)

print(X_train.shape)
```

```python
# Perform StandarScalar
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train_std = sc.fit_transform(X_train)
X_test_std = sc.transform(X_test)
```

```python

# Recursive Feature Selection
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import RFE
lr = LogisticRegression(solver='lbfgs', random_state=123, max_iter=1000)
rfe = RFE(estimator=lr, n_features_to_select=5, step=1)
rfe.fit(X_train_std, y_train)
X_train_sub = rfe.transform(X_train_std)
```

```python
# Which column selected
rfe.support_
Wine_data.columns[1:][rfe.support_]
```

```python
# RFE as Part of a Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
pipe = make_pipeline(RFE(estimator=lr, step=1),
                     KNeighborsClassifier())
parameters = {'rfe__n_features_to_select': range(1, 13), 
              'kneighborsclassifier__n_neighbors': range(1, 10) }

grid = GridSearchCV(pipe, param_grid=parameters, cv=10, n_jobs=-1)
grid.fit(X_train_std, y_train)

print('Best params:', grid.best_params_)
print('Best accuracy:', grid.best_score_)
```

```python
# Reduced feature set from grid search

grid.best_estimator_.score(X_test_std, y_test)
```

```python
# Full feature set for reference

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_std, y_train)
knn.score(X_test_std, y_test)
```

```python
# Embedded:
# Embedded feature selection means the model selects important features automatically while training.
# It is called embedded because the selection of important features is built into (embedded in) the machine learning algorithm itself.
# Embedded method trains the model and selects important features at the same time.
# Embedded methods are used in models such as:
    # Lasso Regression
    # Ridge Regression
    # Elastic Net
    # Decision Tree
    # Random Forest
    # Gradient Boosting
```

```python
# Step-by-step process

# Model starts training with all features
# Model evaluates contribution of each feature
# Less important features get low weight or zero weight
# Important features get higher weight
# Final model keeps only the useful features
```

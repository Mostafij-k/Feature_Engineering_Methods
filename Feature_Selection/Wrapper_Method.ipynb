```python
# Wrapper methods are a type of feature selection methods that involve using a predictive model(ML model) to score the combination of features.

# It create a subset of input feature from orginal feature than try ML algorithm and find r2/accuracy score. Based on r2 score it select best feature.

# They are called 'Wrapper' method because they 'wrap' this type of model based evaluation around the feature selection process.

# Step of wrapper method
    # 1. Subset Generation
    # 2. Subset Evaluation
    # 3. Stoping Criterion
```

```python
# Type of feature selection
    # Exhaustive feature selection
    # Forward selection
    # Backward elimination
    # Recursive feature elimination
```

```python
# Exhaustive feature selection/Best subset selection
    # Try out all the subset combination
    # Example:
    # f1,f2,y: try: f1-y, f2-y, f1,f2-y : find score and select best subset combination.
```

```python
# Disadvantages:
    # 1. Computational Complexity: Need 2^n-1 evaluation
    # 2. Risk of overfitting
    # 3. Require good evaluation metric

# Note: It is best for low number of feature dataset.
# It is better to use filter method than use wrapper method
```

```python
# Sequential Backward Selection(SBS):
    # First train the model on whole dataset
    # Then Eliminate column one by one and train model
# Start → All Features → Remove one by one → Best subset
# This model is more faster
# Disadvantages:
# Can miss best feature
# Iteration will be the number of feature
```

```python
# Step-by-Step Process
# F={X1​,X2​,X3​,X4​,X5​}
# Train a model using all features.
# Remove X1......=F1 score
# Remove X2.......= f1 score
#.......
```

```python
# Sequential Forward Selection(SFS):
# Sequential Forward Selection starts with no features.
# Adds one feature at a time that improves the model performance the most.
# Start → No Features → Add one by one → Best subset
```

```python
# Embedded:
# Embedded feature selection means the model selects important features automatically while training.
# It is called embedded because the selection of important features is built into (embedded in) the machine learning algorithm itself.
# Embedded method trains the model and selects important features at the same time.
# Embedded methods are used in models such as:
    # Lasso Regression
    # Ridge Regression
    # Elastic Net
    # Decision Tree
    # Random Forest
    # Gradient Boosting
```

```python
# Step-by-step process

# Model starts training with all features
# Model evaluates contribution of each feature
# Less important features get low weight or zero weight
# Important features get higher weight
# Final model keeps only the useful features
```

```python
            #-----Feature Scaling-------#
# Feature scaling is a technique  to standardize the independent features present in the data in a fixed range.
# Bring all input independent features to a similar numerical range or scale.
# Prevent dominance of large-Scale feature.
# This is the last step among all feature engineering process.
```

```python
# Why do we need Feature Scaling?
    # Ensures fair contribution of features
    # Required for distance-based methods
    # Improves model convergence 
    # Without scaling ML algorithms do not perform well
```

```python
# Types of Scaling:
    # 1. Standardization
    # 2. Normalization
```

```python
# Standardization: Also called as Z score normalization.
# Formula of Standardization: Xi-Mean/Standard Deviation 
# After scaling mean of scaled data will be 0
# After scaling SD of scaled data will be 1

# Standard deviation: measures the spread or dispersion of a dataset relative to its mean.
    # Low standard deviation → data points are close to the mean
    # High standard deviation → data points are widely spread out
```

```python
            # -------Standardization Example-------#
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Import Dataset
Data=pd.read_csv('Social_Network_Ads.csv')
Data.shape
```

```python
# Check the dataset
Data.head()
```

```python
# Drop first two column
Data=Data.iloc[:,2:]
```

```python
# Again Check the data
Data.head()
```

```python
# It is recommended that before scaling, devide the dataset into train test split.
    # Train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(
    Data.drop('Purchased', axis=1),
    Data['Purchased'],
    test_size=0.3,
    random_state=0
    )
```

```python
# Shape of X_train
X_train.shape
```

```python
# Shape of X_test
X_test.shape
```

```python
# Now perform standardization
from sklearn.preprocessing import StandardScaler

# Create object of StandarScaler class
scaler=StandardScaler()

# fit the scaler to the train set, it will leran the parameters(mean, SD)
# perform fit only with train set not test_set, it prevent data leakage

scaler.fit(X_train)
# Transform train and test set
X_train_scaled=scaler.transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
# type of X_train, X_test
type(X_train_scaled)  # X_train_scaled loses column names
type(X_test_scaled)  # X_test_scaled loses column names
```

```python
# Convert the numpy array to pandas dataframe
X_train_scaled=pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled=pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

```python
# Check X_train_scaled and X_test_scaled
X_train_scaled
X_test_scaled
```

```python
# Compare original X_train and Scaled X_train
np.round(X_train.describe(), 1)     # Age and salary in different scale
```

```python
np.round(X_train_scaled.describe(), 1)   # Age and Salary in same scale
```

```python
# Show the Effect of Scalling
fig, (ax1,ax2)=plt.subplots(ncols=2, figsize=(12,5))

ax1.scatter(X_train['Age'], X_train['EstimatedSalary'])
ax1.set_title('Before Scaling')
ax2.scatter(X_train_scaled['Age'], X_train_scaled['EstimatedSalary'], color='red')
ax2.set_title('After Scaling')
```

```python
# There is no difference in shape
# But after scaling data becomes mean centric
```

```python
# Draw KDEplot to show the effect of scaling
fig, (ax1, ax2)=plt.subplots(ncols=2, figsize=(12,5))

# Before Scaling
ax1.set_title('Before Scaling')
sns.kdeplot(X_train['Age'], ax=ax1)
sns.kdeplot(X_train['EstimatedSalary'], ax=ax1)

# After Scaling
ax2.set_title('After Standard Scaling')
sns.kdeplot(X_train_scaled['Age'], ax=ax2)
sns.kdeplot(X_train_scaled['EstimatedSalary'], ax=ax2)
```

```python
# After Standard Scaling both feature is comparable
```

```python
# Individual Feature comparison 
fig,(ax1, ax2)=plt.subplots(ncols=2, figsize=(12,5))

# Before scaling
ax1.set_title('Age Distribution Before Scaling')
sns.kdeplot(X_train['Age'], ax=ax1)
# After Scaling
ax2.set_title('Age Distribution After Scaling')
sns.kdeplot(X_train_scaled['Age'], ax=ax2)
```

```python
# The shape of distribution in both (after scaling and before scaling) is equal.
```

```python
# Why Scaling is important?
    # Makes all features comparable
    # Prevents large-value features from dominating
    # Helps models learn faster and better
    # Improves accuracy for distance-based and gradient-based models
```

```python
# Prove by lgistic regression
# Import liraries
from sklearn.linear_model import LogisticRegression

# Create object
lr=LogisticRegression()
lr_scaled=LogisticRegression()

# Fit X_train and X_train_scaled
lr.fit(X_train, y_train)
lr_scaled.fit(X_train_scaled, y_train)
```

```python
# Make prediction
y_pred=lr.predict(X_test)
y_pred_scaled=lr_scaled.predict(X_test_scaled)
```

```python
# Show the model accuracy
from sklearn.metrics import accuracy_score
print('Actual:', accuracy_score( y_test, y_pred))
print('Scaled:', accuracy_score( y_test, y_pred_scaled))
```

```python
# Predic with decision tree
from sklearn.tree import DecisionTreeClassifier

# Create object
dt=DecisionTreeClassifier()
dt_scaled=DecisionTreeClassifier()
```

```python
# Train the model
dt.fit(X_train, y_train)
dt_scaled.fit(X_train_scaled, y_train)
```

```python
# Make prediction
y_pred=dt.predict(X_test)
y_pred_scaled=dt.predict(X_test_scaled)
```

```python
# Check the accuracy
print('Actual:', accuracy_score(y_test, y_pred))
print('Scaled:', accuracy_score(y_test, y_pred_scaled))
```

```python
# When to use Standardization
    # K-Means: Use the euclidian distnace measure
    # K-Nearest-Neighbours: Measure the distance between pairs of samples.
    # PCA   
    # ANN : Apply Gradient descent
    # Gradient Discent
```

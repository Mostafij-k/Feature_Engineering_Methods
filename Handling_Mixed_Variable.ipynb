```python
# When numerical + categorical (other) variables are used together, they’re called mixed variables.
```

```python
# Import required variables
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Import Dataset
df = pd.read_csv('Mixed_Variable.csv')
```

```python
# Check Dataset
df.head(2)
```

```python
# Draw bar plot to know the frequency
plt.figure(figsize=(3,3))
ax = df['number'].value_counts().plot.bar()
ax.set_title('Passengers travelling with')
plt.show()
```

```python
# extract numerical part
df['number_numerical'] = pd.to_numeric(
    df['number'],
    errors='coerce',
    downcast='integer')
```

```python
# extract categorical part
df['number_categorical'] = np.where(
    df['number_numerical'].isnull(),
    df['number'],np.nan)
```

```python
# MICE/Iterative Imputer:
    # MICE: Multivariate Imputation by Chained Equations
    # It will be used when Data is Missing at Random(MAR) means missing value depends on ther column
# Advantages:
    # More accurate
# Disadvantages
    # Required more memory
    # Slow
```

```python
# How it works
    # First fill all the missing value with the mean value of corresponding column
    # Then, remove missing value of col1 
    # Predict col1 missing value using other cols
    # Then, remove missing value of col2 
    # Predict col1 missing value using other cols
    # Findout difference = iteration0-iteration1
    # Iteration will be continue until difference become 0
```

```python
# Import requires libraries
import pandas as pd
import numpy as np

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.preprocessing import LabelEncoder
```

```python
Data=pd.read_csv('titanic.csv', usecols=['Age', 'Cabin', 'Embarked'])
Data.head()
```

```python
# 
Data.isnull().sum()
```

```python
# Perform label encoder
le_embarked = LabelEncoder()
le_cabin = LabelEncoder()

Data['Embarked'] = Data['Embarked'].astype(str)
Data['Cabin'] = Data['Cabin'].astype(str)

Data['Embarked'] = le_embarked.fit_transform(Data['Embarked'])
Data['Cabin'] = le_cabin.fit_transform(Data['Cabin'])
```

```python
# Perform Iterative Imputer
imputer = IterativeImputer(max_iter=10, random_state=42)
imputed_data = imputer.fit_transform(Data)
df_imputed = pd.DataFrame(imputed_data, columns=Data.columns)

print(df_imputed.head())
```
